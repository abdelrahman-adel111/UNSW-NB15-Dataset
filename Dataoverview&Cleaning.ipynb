### Type Handeling

```python
def clean_and_map_type(t):
    t_str = str(t).strip().lower()
    if 'integer' in t_str or 'binary' in t_str:
        return 'Int64'
    elif 'float' in t_str:
        return 'float64'
    elif 'timestamp' in t_str or 'date' in t_str:
        return 'datetime64[ns]'
    else:
        return 'object'

type_dict = {
    col: clean_and_map_type(t) 
    for col, t in zip(Columns, feature['Type '])
}

for col, target_type in type_dict.items():
    if target_type == 'Int64':
        DataSet[col] = pd.to_numeric(DataSet[col], errors='coerce').astype('Int64')
        
    elif target_type == 'float64':
        DataSet[col] = pd.to_numeric(DataSet[col], errors='coerce')
        
    elif target_type == 'datetime64[ns]':
        DataSet[col] = pd.to_datetime(DataSet[col], errors='coerce')
        
    else:
        DataSet[col] = DataSet[col].astype(str)

DataSet.info()
```

### Checking Outliers

```python
def FindAndHandelQutlierInData(data, col):
    clean_col = data[col].dropna()
    Q1 = np.quantile(clean_col, .25)
    Q3 = np.quantile(clean_col, .75)
    IQR = Q3 - Q1
    lower = max(0, Q1 - 1.5 * IQR)
    upper = Q3 + 1.5 * IQR

    NoOutlier = ((clean_col > upper) | (clean_col < lower)).sum()
    print(f"{col:<19}: found {NoOutlier:<9} outlier(s) -> bounds = [{lower:.2f}, {upper:.2f}]")

for col in DataSet.columns:
    if DataSet[col].dtype in ['Int64', 'float64']:
        FindAndHandelQutlierInData(DataSet, col)
```

Here is a brief summary of the preprocessing steps and reasons:

* **Fixed Encoding**: Used `latin1` to resolve reading errors and removed byte order mark (BOM) characters (`ï»¿`).
* **Dropped Duplicates**: Removed identical rows to prevent model overfitting and data leakage.
* **Cleaned Anomalous Values**: Converted `is_ftp_login` to a clean binary flag (`0`/`1`), replaced `'-'` in `service` with `'none'`, and merged `'Backdoor'` into `'Backdoors'` in `attack_cat`.
* **Kept Outliers**: Preserved extreme values because high packet/byte rates represent key attack signatures.
* **Dropped IP Addresses**: Removed `srcip` and `dstip` so the model learns attack behaviors rather than memorizing specific network addresses.

## Dataset Overview — UNSW-NB15

The UNSW-NB15 dataset was created by the Cyber Range Lab of the Australian
Centre for Cyber Security (ACCS) using the IXIA PerfectStorm tool to generate
a hybrid of real modern normal network traffic and synthetic contemporary
attack behaviours.

- **Total records used in this project:** ~2.5M raw connections (4 CSV parts
  combined: `UNSW-NB15_1.csv` → `UNSW-NB15_4.csv`)
- **Features:** 49 columns (47 predictive features + 2 target columns)
- **Feature reference:** `NUSW-NB15_features.csv` (name, type, description
  for every column — no headers exist in the raw CSVs, so this file is
  required to reconstruct them)

### Feature groups
| Group | Examples | What it captures |
|---|---|---|
| Flow identifiers | `srcip`, `sport`, `dstip`, `dsport`, `proto` | Connection endpoints & protocol |
| Basic features | `dur`, `sbytes`, `dbytes`, `sttl`, `dttl`, `state` | Duration, byte counts, TTLs, connection state |
| Content features | `service`, `swin`, `dwin`, `stcpb`, `dtcpb`, `smeansz`, `dmeansz` | Payload/header-level details |
| Time features | `Sjit`, `Djit`, `Sintpkt`, `Dintpkt`, `tcprtt`, `synack`, `ackdat` | Jitter and round-trip timing |
| Connection-count (ct_*) features | `ct_srv_src`, `ct_dst_ltm`, `ct_dst_src_ltm`, etc. | Aggregated counts of similar connections in a time window (useful for detecting scans/floods) |
| Targets | `Label`, `attack_cat` | Binary and multiclass ground truth |

### Target variables
- **`Label`** (binary): `0` = Normal traffic, `1` = Attack
- **`attack_cat`** (multiclass): the specific attack family when `Label == 1`
  — Fuzzers, Analysis, Backdoors, DoS, Exploits, Generic, Reconnaissance,
  Shellcode, Worms. Empty/NaN when the record is Normal.

## Data Quality Issues Found & How They Were Handled

| Issue | Where | Fix Applied | Why |
|---|---|---|---|
| No column headers in raw CSVs | All 4 files | Column names/types mapped from `NUSW-NB15_features.csv` | The dataset is split into 4 raw parts without headers |
| Encoding errors | `NUSW-NB15_features.csv` | Read with `ISO-8859-1` / `latin1` | UTF-8 failed on some special characters in feature descriptions |
| Type mismatches | Almost every column | Explicit casting based on the feature file's declared type (`Int64`, `float64`, `datetime64[ns]`, `object`) | Raw CSVs load everything as generic objects; correct dtypes are required for numeric analysis and modeling |
| Dirty binary flag | `is_ftp_login` | Any non-zero/invalid value forced to `1`, missing forced to `0` | Column is supposed to be strictly binary (0/1) but contained inconsistent values |
| Inconsistent categorical labels | `attack_cat` | `NaN` → `'Normal'`, `'Backdoor'` → `'Backdoors'` (typo fix) | NaN only occurs for benign traffic and needs an explicit label for multiclass modeling; `'Backdoor'`/`'Backdoors'` was a spelling inconsistency in the raw data that would otherwise count as two different classes |
| Placeholder value | `service` | `'-'` → `'none'` | `'-'` meant "no service detected", clearer as an explicit category |
| Leaky identifiers | `srcip`, `dstip` | Dropped | Using raw IP addresses lets the model memorize specific known-attacker addresses instead of learning generalizable attack *behavior* — this would fail on any new/unseen network |
| Leading/trailing whitespace | All `object` columns (e.g. `' Fuzzers'` vs `'Fuzzers'`) | Stripped with `.str.strip()` | Would otherwise be silently counted as separate categories |
| Fully-empty rows/columns | Whole dataset | Dropped | No informational value |
| Duplicate records | Whole dataset | Dropped (~exact duplicate rows) | Prevents the model from overweighting repeated connections and reduces train/test leakage risk |
| Outliers (IQR method) | All numeric features | **Kept, not removed** | In network security, extreme values (huge byte counts, unusual durations, packet floods) are often *the actual attack signature* — removing them would strip out the very patterns the model needs to learn |
